```python
import pandas as pd

from pathlib import Path

import numpy as np
import os
import json
```

```python
class PropertiesNMR:
    """Properties from JSON file from Orca NMR calculations"""

    def __init__(self, path):
        """this class reads an orca json file from a given path and extracts properties
        Args:
            path (Path): Path to the orca property.json file
        The class is initialized by reading the json file
        """
        
        self.path = path

        self.results = {}

        if 'nmr.property.json' in os.listdir(path):
            self.nmrjson = json.load(open(path + 'nmr.property.json'))
        else:
            print("No nmr.property.json file found in the specified path. Results dictionary will be empty.")
            self.nmrjson = {}

        # initialize lists for atoms, bonds, angles, dihedrals
        self.atom_list = None

    def define_atom_list(self, atom_list):
        self.atom_list = atom_list



    #-- Saniso calculation --#
    def get_saniso(self):
        """
        """
        if self.atom_list is None:
            print("Atom list is not defined. Please define it using define_atom_list().")
            return
        
        for at_idx in self.atom_list:
            self.results[f"saniso_{at_idx}"] = self.nmrjson['Geometries'][-1]['Chemical_Shift'][0]['saniso'][at_idx-1]


    def get_siso(self):
        """
        """
        if self.atom_list is None:
            print("Atom list is not defined. Please define it using define_atom_list().")
            return

        for at_idx in self.atom_list:
            self.results[f"siso_{at_idx}"] = self.nmrjson['Geometries'][-1]['Chemical_Shift'][0]['siso'][at_idx-1]
```

```python
P = PropertiesNMR(path='orca_example/')
P.get_saniso()
P.define_atom_list([1, 2])
P.get_siso()
P.results
```
