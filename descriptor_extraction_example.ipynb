```python
import orca_descriptors as od
```

```python
# descriptors extraction for one orca job below
# extraction of atomic and global electron density derived parameters (charges, energies, NMR...)
```

```python
# specify parameters for the OrcaReader class (calls the init function)
singlepoint = od.OrcaReader(directory       = "orca_exampleNi_2",   #<- this is the folder with the job outputs
                            basename        = "singlepoint",        #<- this is the type of job outputs the parameters will be extracted from (to extract descriptors for the json summary file)
                            out_file_path   = "orca_exampleNi_2/calcParam_initialOpt.out") #<- this is the complete ORCA output file path

# define atom to extract descriptors from
singlepoint.set_atoms({1:"H1",2:"Ni2",3:"H3"})  #<- define the atom numbers to extract atomic descriptors from (Warning: this function takes atoms numbered on a 1-based index, ORCA uses a 0-based index)

# use the class functions to extract descriptors
singlepoint.parse_property_json()               #<- loads the json file associated with the basename 
singlepoint.parse_gbw_json()                    #<- loads the 

singlepoint.get_all_charges()                   #<- extract all charges
singlepoint.get_homolumo()                      #<- extract homo-lumo   (to fix)
singlepoint.get_scfenergy()                     #<- extract scf energy

singlepoint.results
```

```python
# extraction of descriptors from the frequency job (Gibbs Free Energy and Enthalpy)

frequency = od.OrcaReader(directory         = "orca_exampleNi_2",   #<- this is the folder with the job outputs
                            basename        = "frequency",        #<- this is the type of job outputs the parameters will be extracted from (to extract descriptors for the json summary file)
                            out_file_path   = "orca_exampleNi_2/calcParam_initialOpt.out") #<- this is the complete ORCA output file path


# use the class functions to extract descriptors
frequency.parse_property_json()               #<- loads the json file associated with the basename 

# get enthalpy and Gibbs Free Energy
frequency.get_enthalpy()
frequency.get_gfe()

# TO DO: add a function that checks that all frequencies are positive
frequency.results
```

```python
# extraction of descriptors from the nmr job

nmr = od.OrcaReader(directory       = "orca_exampleNi_2",   #<- this is the folder with the job outputs
                    basename        = "nmr",        #<- this is the type of job outputs the parameters will be extracted from (to extract descriptors for the json summary file)
                    out_file_path   = "orca_exampleNi_2/calcParam_initialOpt.out") #<- this is the complete ORCA output file path

# define atom to extract descriptors from
nmr.set_atoms({1:"H1",2:"Ni2",3:"H3"})  #<- define the atom numbers to extract atomic descriptors from (Warning: this function takes atoms numbered on a 1-based index, ORCA uses a 0-based index)

# use the class functions to extract descriptors
nmr.parse_property_json()               #<- loads the json file associated with the basename 


# get iso and aniso NMR shifts 
nmr.get_siso()
nmr.get_saniso()

# TO DO: add a function that checks that all frequencies are positive
nmr.results
```

```python
# extraction of descriptors from the NBO job: NBOProps class
nbo = od.NBOProps("orca_exampleNi_2/calcParam_initialOpt.out") 
nbo.get_NPA()
nbo.get_NAO()
nbo.get_NBO_summary()
nbo.get_orbital_details()
nbo.get_NHO_directionality()
nbo.get_second_order_perturbation()
nbo.get_cycle_info()

nbo.results #<- this file has a lot of information that we need to reduce to the atoms and bonds we want to analyze only.

# TO DO:
# add a specification for atoms and bonds (pairs of atoms numbers) we want to select information for and retrived them from the results to get just the descriptors needed.
```

```python
# getting the steric descriptors
P = od.PropertiesXYZ(path="orca_exampleNi_2/")

# reset the xyz.json file (optionnal)
P.reset_results()

# define atomic inputs
P.define_atom_list([1, 2])
P.define_bond_list([[1, 2], [2, 3], [1, 3]])
P.define_angle_list([[1, 2, 3], [1, 2, 4], [1, 2, 5], [3, 2, 5]])
P.define_dihedral_list([[1, 2, 3, 4], [2, 4, 14, 15], [5, 4, 14, 15]])
P.quadrant_list = [[2, 1, 3, 4, 6]]  # example: [metal center atom (idx 0), z-axis defined by two atoms (idxs 1 and 2), xz-plane defined by two more atoms (idxs 3 and 4)

# collect properties
P.get_Vbur()
P.get_bond_distances()
P.get_Sterimol_parameters()
P.get_angles()
P.get_dihedrals()
P.get_vbur_octants()
P.get_vbur_quadrants()

P.save_results_to_json()
P.results
```

```python
# concatenate all descriptors

all_descriptors = {**P.results, **nbo.results, **frequency.results, **singlepoint.results, **nmr.results}

print(len(all_descriptors))
```
